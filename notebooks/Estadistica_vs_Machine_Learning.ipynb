Aunque sabemos que esta situación rara vez ocurre en la realidad, resulta ideal para comprender cómo responde la estadística tradicional cuando las relaciones entre las variables son prácticamente lineales.

Este primer experimento servirá como punto de partida.

## Experimento 1

Para este primer escenario construiremos un conjunto de datos sintéticos donde el rendimiento del cultivo dependerá casi exclusivamente de la cantidad de lluvia recibida.

Supondremos que:

- existe una relación prácticamente lineal entre ambas variables;
- el resto de los factores permanecen constantes;
- únicamente incorporaremos un pequeño nivel de variabilidad aleatoria para representar las diferencias naturales que siempre existen en cualquier sistema biológico.

Este será el escenario más sencillo de todo el proyecto.

Precisamente por ello, esperamos que un modelo de Regresión Lineal describa adecuadamente el comportamiento del cultivo.

## ¿Qué esperamos encontrar?

Antes de ejecutar cualquier modelo, vale la pena formular una hipótesis.

Si el rendimiento depende casi exclusivamente de la lluvia, es razonable esperar que:

- exista una correlación positiva elevada entre ambas variables;
- una Regresión Lineal represente adecuadamente el comportamiento del cultivo;
- los modelos de Machine Learning no ofrezcan una mejora significativa respecto a la estadística tradicional.

En otras palabras, este primer experimento nos permitirá comprobar que, cuando la realidad es sencilla, la estadística continúa siendo una herramienta extraordinariamente eficaz.

```python
# ==========================================================
# EXPERIMENTO 1
# ¿Qué ocurre cuando una sola variable explica casi todo
# el rendimiento del cultivo?
# ==========================================================

# En este escenario construiremos un conjunto de datos
# sintéticos donde la precipitación será la principal
# variable explicativa del rendimiento.
#
# El objetivo NO es representar un cultivo real.
#
# El objetivo es construir un problema sencillo que permita
# observar el comportamiento de la estadística tradicional.
```

```python
# ==========================================================
# Generación del Dataset Sintético - Experimento 1
# ==========================================================
#
# En este primer experimento simularemos 1.000 campañas
# agrícolas independientes.
#
# El supuesto principal es muy sencillo:
#
# El rendimiento del cultivo depende casi exclusivamente
# de la cantidad de precipitación recibida durante la campaña.
#
# Para representar la variabilidad natural que existe en
# cualquier sistema biológico, incorporaremos un pequeño
# componente aleatorio (ruido).
#
# Este conjunto de datos NO representa un cultivo específico.
# Su objetivo es únicamente ilustrar un escenario donde una
# relación lineal describe adecuadamente el fenómeno.
# ==========================================================

import numpy as np
import pandas as pd

# ----------------------------------------------------------
# Garantizar reproducibilidad del experimento
# ----------------------------------------------------------

np.random.seed(42)

# ----------------------------------------------------------
# Número de campañas agrícolas simuladas
# ----------------------------------------------------------

n = 1000

# ----------------------------------------------------------
# Variable independiente:
# Precipitación acumulada durante la campaña (mm)
# ----------------------------------------------------------

precipitacion = np.random.uniform(
    low=300,
    high=1200,
    size=n
)

# ----------------------------------------------------------
# Variabilidad natural del sistema
# ----------------------------------------------------------

ruido = np.random.normal(
    loc=0,
    scale=0.8,
    size=n
)

# ----------------------------------------------------------
# Modelo conceptual
# ----------------------------------------------------------

rendimiento = (
    2
    + 0.008 * precipitacion
    + ruido
)

# ----------------------------------------------------------
# Construcción del Dataset
# ----------------------------------------------------------

df_exp1 = pd.DataFrame({
    "Precipitacion_mm": precipitacion,
    "Rendimiento_t_ha": rendimiento
})

# ----------------------------------------------------------
# Resumen del experimento
# ----------------------------------------------------------

print("=" * 60)
print("EXPERIMENTO 1 - DATASET SINTÉTICO")
print("=" * 60)

print(f"Campañas agrícolas simuladas : {len(df_exp1):,}")
print(f"Variables                    : {df_exp1.shape[1]}")

print("\nPrimeras observaciones:\n")

display(df_exp1.head())
```

## ¿Qué acabamos de construir?

En este primer experimento hemos generado **1.000 campañas agrícolas simuladas**, donde cada observación representa una combinación de precipitación y rendimiento del cultivo.

A simple vista podemos notar dos aspectos importantes:

- La **precipitación** presenta valores diferentes en cada campaña, simulando la variabilidad climática que puede existir entre distintas temporadas agrícolas.
- El **rendimiento** también varía entre campañas, aunque sigue una tendencia relacionada con la cantidad de lluvia recibida.

En este escenario todavía no intervienen otros factores como la temperatura, el tipo de suelo o la fertilización. Hemos construido deliberadamente un sistema muy sencillo donde la precipitación explica casi todo el comportamiento del cultivo.

Este será nuestro punto de partida.

En los siguientes experimentos iremos incorporando nuevas variables e interacciones para observar cómo aumenta la complejidad del problema y cómo cambia el desempeño de los distintos modelos de predicción.

## ¿Qué nos dicen estas estadísticas?

Las estadísticas descriptivas muestran un comportamiento coherente con el experimento que hemos diseñado.

Observamos que la precipitación presenta una amplia variabilidad entre campañas agrícolas, mientras que el rendimiento cambia siguiendo una tendencia similar.

En este escenario no existen valores faltantes ni observaciones atípicas, ya que los datos fueron generados de forma controlada.

Hasta este momento, todo indica que estamos trabajando con un problema relativamente sencillo, donde una única variable parece explicar gran parte del comportamiento del cultivo.

El siguiente paso consistirá en visualizar esa relación para comprobar si realmente existe una tendencia lineal entre la precipitación y el rendimiento.

```python
# ==========================================================
# Figura 1
# Cuando una variable parece explicarlo todo
# ==========================================================

from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# ----------------------------------------------------------
# Preparación de los datos
# ----------------------------------------------------------

X = df_exp1[["Precipitacion_mm"]]
y = df_exp1["Rendimiento_t_ha"]

# ----------------------------------------------------------
# Ajuste del modelo de Regresión Lineal
# ----------------------------------------------------------

modelo_lr = LinearRegression()
modelo_lr.fit(X, y)

# Predicciones de la línea de regresión
pred_lr = modelo_lr.predict(X)

# ----------------------------------------------------------
# Ordenar los datos para dibujar correctamente la línea
# ----------------------------------------------------------

orden = X["Precipitacion_mm"].argsort()

X_ordenado = X.iloc[orden]
pred_ordenada = pred_lr[orden]

# ----------------------------------------------------------
# Construcción del gráfico
# ----------------------------------------------------------

plt.figure(figsize=(10, 6))

# Nube de puntos
plt.scatter(
    df_exp1["Precipitacion_mm"],
    df_exp1["Rendimiento_t_ha"],
    alpha=0.65,
    s=40,
    color="#2E86C1",
    edgecolor="white",
    linewidth=0.4,
    label="Campañas agrícolas simuladas"
)

# Línea de regresión
plt.plot(
    X_ordenado,
    pred_ordenada,
    color="#2E8B57",
    linewidth=3,
    label="Regresión Lineal"
)

# ----------------------------------------------------------
# Personalización
# ----------------------------------------------------------

plt.title(
    "Figura 1. Cuando una variable parece explicarlo todo",
    fontsize=16,
    weight="bold",
    pad=15
)

plt.xlabel(
    "Precipitación acumulada durante la campaña (mm)",
    fontsize=12
)

plt.ylabel(
    "Rendimiento del cultivo (t/ha)",
    fontsize=12
)

plt.grid(True, alpha=0.25)

plt.legend(frameon=True)

plt.figtext(
    0.5,
    -0.03,
    "Datos sintéticos generados en Python • Elaboración propia",
    ha="center",
    fontsize=10,
    color="gray"
)

plt.tight_layout()

plt.show()
```

```python
# ==========================================================
# Evaluación del modelo de Regresión Lineal
# ==========================================================

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

import numpy as np

# ----------------------------------------------------------
# Correlación de Pearson
# ----------------------------------------------------------

corr_pearson = df_exp1["Precipitacion_mm"].corr(
    df_exp1["Rendimiento_t_ha"]
)

# ----------------------------------------------------------
# Métricas del modelo
# ----------------------------------------------------------

r2_lr = r2_score(y, pred_lr)

rmse_lr = np.sqrt(
    mean_squared_error(y, pred_lr)
)

mae_lr = mean_absolute_error(y, pred_lr)

# ----------------------------------------------------------
# Resultados
# ----------------------------------------------------------

print("=" * 60)
print("DESEMPEÑO DEL MODELO - REGRESIÓN LINEAL")
print("=" * 60)

print(f"Correlación de Pearson : {corr_pearson:.4f}")

print(f"R²                    : {r2_lr:.4f}")

print(f"RMSE                  : {rmse_lr:.4f}")

print(f"MAE                   : {mae_lr:.4f}")
```

## ¿Qué nos indican estas métricas?

Los resultados obtenidos confirman lo que habíamos observado visualmente en la Figura 1.

La correlación entre la precipitación y el rendimiento es muy elevada, lo que indica una fuerte asociación lineal entre ambas variables.

Asimismo, el coeficiente de determinación (R²) muestra que el modelo de Regresión Lineal explica prácticamente toda la variabilidad presente en el rendimiento del cultivo.

Los valores reducidos del RMSE y del MAE indican que las diferencias entre los valores observados y las predicciones del modelo son pequeñas.

En conjunto, estas métricas respaldan una conclusión importante:

> **Cuando un fenómeno puede describirse mediante una relación esencialmente lineal, la estadística tradicional continúa siendo una herramienta altamente efectiva para comprender y predecir el comportamiento del sistema.**

Sin embargo, este resultado corresponde a un escenario deliberadamente simple.

## ¿Qué aprendió la Regresión Lineal?

La Regresión Lineal intenta describir la relación entre la precipitación y el rendimiento mediante una ecuación sencilla.

En otras palabras, busca responder una pregunta muy concreta:

> **¿Cómo cambia el rendimiento del cultivo cuando aumenta la precipitación?**

Para ello ajusta una línea recta que representa la tendencia promedio observada en los datos.

Veamos cuál es la ecuación obtenida para este experimento.

```python
# ==========================================================
# Ecuación del modelo de Regresión Lineal
# ==========================================================

pendiente = modelo_lr.coef_[0]
intercepto = modelo_lr.intercept_

print("=" * 60)
print("ECUACIÓN DEL MODELO")
print("=" * 60)

print(f"Intercepto : {intercepto:.4f}")
print(f"Pendiente  : {pendiente:.6f}")

print("\nEcuación estimada:")

print(
    f"Rendimiento = {intercepto:.4f} + "
    f"({pendiente:.6f} × Precipitación)"
)
```

## Una observación interesante

La ecuación obtenida es muy similar a la utilizada para construir el conjunto de datos sintéticos.

Esto era justamente lo que esperábamos.

Recordemos que este primer experimento fue diseñado deliberadamente para representar una relación prácticamente lineal entre la precipitación y el rendimiento.

Por esa razón, la Regresión Lineal consigue recuperar con gran precisión el comportamiento del sistema.

En otras palabras, el modelo está aprendiendo correctamente la lógica con la que fue construido el experimento.

En el experimento anterior construimos un escenario deliberadamente simple.

Los suelos con mayor contenido de materia orgánica suelen presentar mejores condiciones físicas, químicas y biológicas, favoreciendo el desarrollo del cultivo y su capacidad para aprovechar el agua disponible.

En este segundo experimento incorporaremos esta nueva variable.

## Hipótesis

Antes de construir el nuevo conjunto de datos, formularemos nuevamente una hipótesis.

Esperamos que:

- la precipitación continúe influyendo positivamente sobre el rendimiento;
- la materia orgánica también contribuya al incremento de la producción;
- ambas variables actúen de manera independiente;
- la Regresión Lineal siga describiendo correctamente el comportamiento del sistema.

En consecuencia, esperamos que la estadística tradicional continúe ofreciendo un excelente desempeño, aun cuando el problema incorpore más de una variable explicativa.

La verdadera complejidad todavía no ha aparecido.

Eso ocurrirá en los siguientes experimentos.

```python
# ==========================================================
# Generación del Dataset Sintético - Experimento 2
# ==========================================================
#
# En este segundo experimento simularemos nuevamente
# 1.000 campañas agrícolas.
#
# A diferencia del Experimento 1, ahora el rendimiento
# dependerá de tres variables:
#
# - Precipitación
# - Materia orgánica del suelo
# - Temperatura media
#
# Todas ellas actuarán de forma independiente,
# manteniendo una relación esencialmente lineal.
#
# ==========================================================

import numpy as np
import pandas as pd

# ----------------------------------------------------------
# Reproducibilidad
# ----------------------------------------------------------

np.random.seed(42)

# ----------------------------------------------------------
# Número de campañas agrícolas simuladas
# ----------------------------------------------------------

n_campanias = 1000

# ----------------------------------------------------------
# Variables explicativas
# ----------------------------------------------------------

precipitacion = np.random.uniform(
    low=300,
    high=1200,
    size=n_campanias
)

materia_organica = np.random.uniform(
    low=1.0,
    high=5.0,
    size=n_campanias
)

temperatura = np.random.uniform(
    low=15,
    high=30,
    size=n_campanias
)

# ----------------------------------------------------------
# Variabilidad natural
# ----------------------------------------------------------

ruido = np.random.normal(
    loc=0,
    scale=0.8,
    size=n_campanias
)

# ----------------------------------------------------------
# Modelo conceptual
#
# Todas las variables contribuyen al rendimiento
# de manera independiente.
# ----------------------------------------------------------

rendimiento = (
    1.5
    + 0.006 * precipitacion
    + 0.90 * materia_organica
    + 0.18 * temperatura
    + ruido
)

# ----------------------------------------------------------
# Construcción del DataFrame
# ----------------------------------------------------------

df_exp2 = pd.DataFrame({

    "Precipitacion_mm": precipitacion,

    "Materia_Organica_pct": materia_organica,

    "Temperatura_C": temperatura,

    "Rendimiento_t_ha": rendimiento

})

# ----------------------------------------------------------
# Resumen
# ----------------------------------------------------------

print("=" * 65)
print("EXPERIMENTO 2 - DATASET SINTÉTICO")
print("=" * 65)

print(f"Campañas agrícolas simuladas : {len(df_exp2):,}")
print(f"Variables                    : {df_exp2.shape[1]}")

print("\nPrimeras observaciones:\n")

display(df_exp2.head())
```

## ¿Qué cambió respecto al experimento anterior?

En el primer experimento, el rendimiento del cultivo dependía únicamente de la precipitación.

Ahora hemos incorporado dos nuevos factores:

- la materia orgánica del suelo;
- la temperatura media durante la campaña.

Las tres variables contribuyen al rendimiento de manera independiente, manteniendo una relación lineal con la variable objetivo.

Este escenario representa una situación más cercana a la realidad agrícola, donde múltiples factores influyen simultáneamente sobre la producción.

La pregunta que intentaremos responder ahora es:

> **¿Sigue siendo suficiente la Regresión Lineal cuando el fenómeno depende de varias variables, pero todas ellas mantienen relaciones esencialmente lineales?**

## ¿Qué ocurre cuando analizamos todas las variables simultáneamente?

Hasta ahora hemos estudiado el efecto de cada variable por separado.

Sin embargo, en un problema agrícola real, el rendimiento no depende únicamente de un solo factor.

La precipitación, la materia orgánica y la temperatura actúan de manera simultánea.

Por esta razón, construiremos un modelo de **Regresión Lineal Múltiple**, que permitirá estimar el efecto conjunto de las tres variables sobre el rendimiento del cultivo.

La pregunta ahora es:

> **¿Será capaz la Regresión Lineal de mantener su excelente desempeño cuando el fenómeno depende de varias variables al mismo tiempo?**

```python
# ==========================================================
# Regresión Lineal Múltiple
# ==========================================================

from sklearn.linear_model import LinearRegression

# ----------------------------------------------------------
# Variables predictoras
# ----------------------------------------------------------

X = df_exp2[[
    "Precipitacion_mm",
    "Materia_Organica_pct",
    "Temperatura_C"
]]

# ----------------------------------------------------------
# Variable objetivo
# ----------------------------------------------------------

y = df_exp2["Rendimiento_t_ha"]

# ----------------------------------------------------------
# Construcción del modelo
# ----------------------------------------------------------

modelo_lr = LinearRegression()

modelo_lr.fit(X, y)

# ----------------------------------------------------------
# Predicciones
# ----------------------------------------------------------

pred_lr = modelo_lr.predict(X)
```

```python
# ==========================================================
# Coeficientes del modelo
# ==========================================================

coeficientes = pd.DataFrame({

    "Variable": X.columns,

    "Coeficiente": modelo_lr.coef_

})

display(coeficientes.round(4))

print("\nIntercepto:", round(modelo_lr.intercept_,4))
```

## ¿Qué aprendió el modelo?

Los coeficientes estimados son muy próximos a los valores utilizados para construir el conjunto de datos sintéticos.

Este resultado confirma que la Regresión Lineal ha conseguido identificar correctamente la contribución individual de cada variable.

Desde un punto de vista práctico, esto significa que el modelo no solo realiza buenas predicciones, sino que además permite interpretar el efecto de cada factor sobre el rendimiento del cultivo.

Esta es una de las grandes fortalezas de la estadística tradicional cuando las relaciones entre las variables permanecen esencialmente lineales.

```python
# ==========================================================
# Evaluación del modelo de Regresión Lineal Múltiple
# ==========================================================

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

import numpy as np

# ----------------------------------------------------------
# Métricas del modelo
# ----------------------------------------------------------

r2_lr = r2_score(y, pred_lr)

rmse_lr = np.sqrt(
    mean_squared_error(y, pred_lr)
)

mae_lr = mean_absolute_error(y, pred_lr)

# ----------------------------------------------------------
# Tabla resumen
# ----------------------------------------------------------

metricas_exp2 = pd.DataFrame({

    "Métrica": [

        "R²",

        "RMSE",

        "MAE"

    ],

    "Valor": [

        r2_lr,

        rmse_lr,

        mae_lr

    ]

})

print("=" * 65)
print("DESEMPEÑO DEL MODELO - REGRESIÓN LINEAL MÚLTIPLE")
print("=" * 65)

display(metricas_exp2.round(4))
```

## ¿Qué nos indican estas métricas?

Los resultados muestran que la Regresión Lineal Múltiple continúa describiendo el comportamiento del sistema con un alto nivel de precisión.

El coeficiente de determinación (R²) indica que el modelo explica la mayor parte de la variabilidad observada en el rendimiento del cultivo.

Asimismo, los valores reducidos del RMSE y del MAE reflejan que las diferencias entre los valores observados y las predicciones son pequeñas.

En consecuencia, el incremento en el número de variables no ha reducido la capacidad predictiva del modelo.

¿Por qué?

Porque el problema sigue siendo esencialmente lineal.

Cada variable aporta información al rendimiento, pero lo hace de forma independiente y sin generar relaciones complejas.

En estas condiciones, la estadística tradicional continúa ofreciendo un desempeño sobresaliente.

## Una reflexión antes de continuar

Hasta este momento hemos comprobado que la Regresión Lineal responde muy bien en dos escenarios diferentes:

- cuando el rendimiento depende de una sola variable;
- cuando depende de varias variables, siempre que sus efectos sean independientes y aproximadamente lineales.

Esto nos lleva a una pregunta inevitable.

La agricultura es un sistema dinámico donde los factores rara vez actúan de manera aislada.

- el efecto de la fertilización puede variar dependiendo de las condiciones climáticas.

En otras palabras, las variables comienzan a interactuar entre sí.

Y cuando esto ocurre, la realidad deja de comportarse como una simple suma de efectos independientes.

# Capítulo 3

# ¿Qué ocurre cuando las variables comienzan a interactuar?

Hasta este momento hemos trabajado con un supuesto muy conveniente desde el punto de vista estadístico:

Cada variable aportaba información al rendimiento del cultivo de manera independiente.

Sin embargo, la naturaleza rara vez funciona de esa forma.

En un sistema agrícola real, los factores ambientales y del suelo no actúan de manera aislada.

Por el contrario, interactúan constantemente.

Por ejemplo:

- Una misma cantidad de lluvia puede producir rendimientos muy diferentes dependiendo de la materia orgánica presente en el suelo.

- Un suelo rico en materia orgánica puede almacenar mejor el agua disponible y amortiguar los efectos de un período seco.

- La temperatura puede potenciar o limitar el efecto de la precipitación según las condiciones del cultivo.

En otras palabras, el efecto de una variable comienza a depender del valor que toman las demás.

Aquí es donde la realidad empieza a alejarse de un comportamiento estrictamente lineal.

La pregunta que intentaremos responder ahora es:

> **¿Sigue siendo suficiente la Regresión Lineal cuando las variables ya no actúan de forma independiente?**

## Hipótesis del Experimento 3

En los experimentos anteriores asumimos que cada variable contribuía al rendimiento del cultivo de manera independiente.

Sin embargo, esta simplificación rara vez representa lo que ocurre en un sistema agrícola real.

En la práctica, los factores ambientales interactúan continuamente.

Por ejemplo, una determinada cantidad de lluvia puede producir un rendimiento elevado en un suelo rico en materia orgánica, pero un rendimiento mucho menor en un suelo degradado.

En este experimento incorporaremos por primera vez una **interacción entre variables**.

La precipitación continuará siendo importante.

La materia orgánica también.

La temperatura seguirá influyendo sobre el cultivo.

Pero ahora el efecto de la precipitación dependerá parcialmente del contenido de materia orgánica del suelo.

Nuestra hipótesis es la siguiente:

- La Regresión Lineal seguirá capturando una parte importante del comportamiento del cultivo.
- Sin embargo, comenzará a perder capacidad para representar completamente el fenómeno.
- La complejidad del problema aumentará, acercándonos un poco más a las condiciones que encontramos en la agricultura real.

```python
# ==========================================================
# Generación del Dataset Sintético - Experimento 3
# ==========================================================
#
# En este tercer experimento simularemos nuevamente
# 1.000 campañas agrícolas.
#
# A diferencia del experimento anterior, el rendimiento
# ya no dependerá únicamente de la suma de los efectos
# individuales de cada variable.
#
# Incorporaremos una interacción entre la precipitación
# y la materia orgánica.
#
# Conceptualmente, esto representa que un suelo con mayor
# contenido de materia orgánica aprovecha mejor el agua
# disponible para el cultivo.
#
# Este comportamiento es frecuente en sistemas agrícolas
# reales y constituye una primera aproximación a fenómenos
# donde las variables dejan de actuar de forma independiente.
# ==========================================================

import numpy as np
import pandas as pd

# ----------------------------------------------------------
# Reproducibilidad
# ----------------------------------------------------------

np.random.seed(42)

# ----------------------------------------------------------
# Número de campañas agrícolas simuladas
# ----------------------------------------------------------

n_campanias = 1000

# ----------------------------------------------------------
# Variables explicativas
# ----------------------------------------------------------

precipitacion = np.random.uniform(
    low=300,
    high=1200,
    size=n_campanias
)

materia_organica = np.random.uniform(
    low=1.0,
    high=5.0,
    size=n_campanias
)

temperatura = np.random.uniform(
    low=15,
    high=30,
    size=n_campanias
)

radiacion = np.random.uniform(
    low=15,
    high=30,
    size=n_campanias
)

# ----------------------------------------------------------
# Variabilidad natural
# ----------------------------------------------------------

ruido = np.random.normal(
    loc=0,
    scale=0.8,
    size=n_campanias
)

# ----------------------------------------------------------
# Interacción entre variables
#
# Un mayor contenido de materia orgánica mejora la capacidad
# del suelo para aprovechar la precipitación.
# ----------------------------------------------------------

interaccion = (
    0.0015
    * precipitacion
    * materia_organica
)

# ----------------------------------------------------------
# Modelo conceptual
# ----------------------------------------------------------

rendimiento = (

    1.2

    + 0.0055 * precipitacion

    + 0.80 * materia_organica

    + 0.15 * temperatura

    + 0.05 * radiacion

    + interaccion

    + ruido

)

# ----------------------------------------------------------
# Construcción del DataFrame
# ----------------------------------------------------------

df_exp3 = pd.DataFrame({

    "Precipitacion_mm": precipitacion,

    "Materia_Organica_pct": materia_organica,

    "Temperatura_C": temperatura,

    "Radiacion": radiacion,

    "Rendimiento_t_ha": rendimiento

})

# ----------------------------------------------------------
# Resumen del experimento
# ----------------------------------------------------------

print("=" * 65)
print("EXPERIMENTO 3 - DATASET SINTÉTICO")
print("=" * 65)

print(f"Campañas agrícolas simuladas : {len(df_exp3):,}")
print(f"Variables                    : {df_exp3.shape[1]}")

print("\nPrimeras observaciones:\n")

display(df_exp3.head())
```

Sin embargo, existe una diferencia fundamental.

En los experimentos anteriores, cada variable contribuía al rendimiento del cultivo de manera independiente.

Ahora hemos incorporado una interacción entre la precipitación y la materia orgánica.

Esto significa que el efecto de la lluvia ya no es exactamente el mismo para todas las campañas agrícolas.

Un suelo con mayor contenido de materia orgánica puede aprovechar mejor el agua disponible, mientras que un suelo degradado obtiene un beneficio menor, aun cuando ambos reciban la misma cantidad de precipitación.

Este pequeño cambio representa una situación mucho más cercana a la realidad agrícola.

```python
# ==========================================================
# Regresión Lineal Múltiple
# (sin considerar explícitamente la interacción)
# ==========================================================

from sklearn.linear_model import LinearRegression

# ----------------------------------------------------------
# Variables predictoras
# ----------------------------------------------------------

X = df_exp3[[
    "Precipitacion_mm",
    "Materia_Organica_pct",
    "Temperatura_C",
    "Radiacion"
]]

# ----------------------------------------------------------
# Variable objetivo
# ----------------------------------------------------------

y = df_exp3["Rendimiento_t_ha"]

# ----------------------------------------------------------
# Construcción del modelo
# ----------------------------------------------------------

modelo_lr = LinearRegression()

modelo_lr.fit(X, y)

# ----------------------------------------------------------
# Predicciones
# ----------------------------------------------------------

pred_lr = modelo_lr.predict(X)
```

```python
# ==========================================================
# Coeficientes del modelo
# ==========================================================

coeficientes = pd.DataFrame({

    "Variable":X.columns,

    "Coeficiente":modelo_lr.coef_

})

display(
    coeficientes.round(4)
)

print("\nIntercepto:",round(modelo_lr.intercept_,4))
```

```python
# ==========================================================
# Evaluación del modelo
# ==========================================================

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

import numpy as np

# ----------------------------------------------------------
# Métricas
# ----------------------------------------------------------

r2_lr = r2_score(y, pred_lr)

rmse_lr = np.sqrt(
    mean_squared_error(y, pred_lr)
)

mae_lr = mean_absolute_error(y, pred_lr)

# ----------------------------------------------------------
# Tabla resumen
# ----------------------------------------------------------

metricas_exp3 = pd.DataFrame({

    "Métrica":[
        "R²",
        "RMSE",
        "MAE"
    ],

    "Valor":[
        r2_lr,
        rmse_lr,
        mae_lr
    ]

})

print("="*65)
print("DESEMPEÑO DEL MODELO - REGRESIÓN LINEAL")
print("="*65)

display(metricas_exp3.round(4))
```

# Del conocimiento agronómico al modelo matemático

Las reglas definidas en la sección anterior representan el comportamiento esperado del sistema agrícola.

```python
# ==========================================================
# Generación del Dataset Sintético - Experimento 4
# ==========================================================
#
# En este experimento construiremos un sistema agrícola
# inspirado en procesos reales.
#
# El objetivo no consiste en representar un cultivo
# específico, sino simular un escenario donde múltiples
# procesos interactúan de manera simultánea.
#
# ==========================================================

import numpy as np
import pandas as pd

# ----------------------------------------------------------
# Reproducibilidad
# ----------------------------------------------------------

np.random.seed(42)

# ----------------------------------------------------------
# Número de campañas agrícolas simuladas
# ----------------------------------------------------------

n_campanias = 1000

# ----------------------------------------------------------
# Variables climáticas
# ----------------------------------------------------------

precipitacion = np.random.uniform(
    250,
    1200,
    n_campanias
)

temperatura = np.random.uniform(
    12,
    35,
    n_campanias
)

radiacion = np.random.uniform(
    15,
    30,
    n_campanias
)

# ----------------------------------------------------------
# Variables del suelo
# ----------------------------------------------------------

materia_organica = np.random.uniform(
    1,
    6,
    n_campanias
)

ph = np.random.uniform(
    4.8,
    8.2,
    n_campanias
)

# ----------------------------------------------------------
# Variables de manejo
# ----------------------------------------------------------

fertilizacion = np.random.uniform(
    50,
    250,
    n_campanias
)

riego = np.random.uniform(
    0,
    300,
    n_campanias
)

print("="*65)
print("VARIABLES BASE GENERADAS")
print("="*65)

print(f"Campañas simuladas : {n_campanias:,}")

print("\nVariables disponibles:")

print("""
✓ Precipitación
✓ Temperatura
✓ Radiación
✓ Materia Orgánica
✓ pH
✓ Fertilización
✓ Riego
""")
```

# Primer paso: definir el contexto del sistema

Las variables anteriores representan algunos de los principales factores que condicionan el rendimiento de un cultivo.

Sin embargo, ninguna de ellas determina por sí sola la producción agrícola.

El rendimiento surge de la interacción entre estas variables.

Por esta razón, antes de calcular la producción del cultivo construiremos varios procesos intermedios que representarán el funcionamiento simplificado del sistema agrícola.

En otras palabras, no modelaremos directamente el rendimiento.

```python
# ==========================================================
# Proceso 1
# Disponibilidad de agua
# ==========================================================
#
# La disponibilidad de agua depende principalmente
# de la precipitación y del riego suplementario.
#
# Se asume que el riego compensa parcialmente
# los déficits de precipitación.
# ==========================================================

agua_disponible = (
    precipitacion
    + 0.60 * riego
)
```

## Proceso 1: Disponibilidad de agua

La lluvia constituye la principal fuente de agua para el cultivo.

Sin embargo, en numerosos sistemas agrícolas el riego suplementario permite reducir parcialmente los efectos de un déficit hídrico.

Por esta razón, en este experimento la disponibilidad de agua será el resultado combinado de ambos componentes.

Este proceso representa una simplificación del balance hídrico que normalmente condiciona el desarrollo del cultivo.

## Proceso 2: Eficiencia en el aprovechamiento del agua

No toda el agua disponible es utilizada de la misma manera por el cultivo.

La capacidad del suelo para almacenar y suministrar agua depende, entre otros factores, de su contenido de materia orgánica.

Los suelos con mayor materia orgánica presentan una mejor estructura, mayor porosidad y una mayor capacidad para retener humedad.

Como consecuencia, una misma cantidad de agua puede producir respuestas diferentes según las condiciones del suelo.

En este experimento representaremos este comportamiento mediante un índice simplificado de eficiencia hídrica.

```python
# ==========================================================
# Proceso 2
# Eficiencia en el aprovechamiento del agua
# ==========================================================

# La materia orgánica mejora la estructura del suelo
# y aumenta la eficiencia con la que el cultivo utiliza
# el agua disponible.

eficiencia_hidrica = (

    1

    + 0.20 * np.log1p(materia_organica)

)

print("="*60)
print("EFICIENCIA HÍDRICA")
print("="*60)

print(f"Valor mínimo : {eficiencia_hidrica.min():.3f}")
print(f"Valor medio  : {eficiencia_hidrica.mean():.3f}")
print(f"Valor máximo : {eficiencia_hidrica.max():.3f}")
```

## Proceso 3: Temperatura óptima para el desarrollo del cultivo

La temperatura es uno de los factores más importantes que condicionan el crecimiento de un cultivo.

Sin embargo, su efecto rara vez es lineal.

Cada especie posee un intervalo óptimo donde los procesos fisiológicos alcanzan su máximo rendimiento.

Cuando la temperatura se aleja de ese rango, el crecimiento disminuye progresivamente.

Por esta razón, en lugar de utilizar una relación lineal, representaremos este comportamiento mediante una función que alcanza su máximo alrededor de una temperatura óptima.

```python
# ==========================================================
# Proceso 3
# Efecto de la temperatura
# ==========================================================
#
# Se asume que el cultivo presenta un rendimiento
# máximo alrededor de los 24 °C.
#
# A medida que la temperatura se aleja de dicho valor,
# el efecto positivo disminuye progresivamente.
# ==========================================================

temperatura_optima = 24

efecto_temperatura = np.exp(

    -((temperatura - temperatura_optima) ** 2)

    / (2 * 5 ** 2)

)
```

>
> Temperaturas inferiores o superiores a ese rango generan distintos tipos de estrés, reduciendo el rendimiento potencial.

## Proceso 4: Disponibilidad de nutrientes en función del pH

El pH del suelo condiciona la disponibilidad de numerosos nutrientes esenciales para el crecimiento de los cultivos.

Cuando el suelo presenta valores excesivamente ácidos o alcalinos, disminuye la disponibilidad de algunos elementos minerales y aumenta el riesgo de deficiencias nutricionales.

Por esta razón, muchos cultivos presentan un intervalo de pH donde alcanzan su mayor potencial productivo.

En este experimento representaremos este comportamiento mediante una respuesta no lineal con un rango óptimo alrededor de pH 6.5.

```python
# ==========================================================
# Proceso 4
# Disponibilidad de nutrientes (efecto del pH)
# ==========================================================
#
# Se considera que la máxima disponibilidad de nutrientes
# ocurre alrededor de un pH cercano a 6.5.
#
# A medida que el pH se aleja de este valor,
# disminuye progresivamente la eficiencia nutricional.
# ==========================================================

ph_optimo = 6.5

efecto_ph = np.exp(

    -((ph - ph_optimo) ** 2)

    / (2 * 0.8 ** 2)

)
```

## Proceso 5: Respuesta del cultivo a la fertilización

La fertilización constituye uno de los principales factores de manejo que influyen sobre el rendimiento agrícola.

Sin embargo, su efecto no es ilimitado.

Cuando el suelo presenta una baja disponibilidad de nutrientes, pequeñas aplicaciones de fertilizante generan incrementos importantes en la producción.

A medida que aumenta la fertilización, el beneficio adicional comienza a disminuir hasta alcanzar un punto donde nuevas aplicaciones producen mejoras cada vez menores.

Este comportamiento se conoce como **Ley de los Rendimientos Decrecientes** y representa uno de los principios más importantes de la producción agrícola.

```python
# ==========================================================
# Proceso 5
# Respuesta del cultivo a la fertilización
# ==========================================================
#
# La fertilización incrementa el rendimiento,
# pero con rendimientos decrecientes.
#
# ==========================================================

efecto_fertilizacion = np.log1p(fertilizacion)
```

> ## 🌱 Interpretación agronómica
>
> La respuesta de un cultivo a la fertilización no es proporcional durante todo el rango de aplicación.
>
> En condiciones de baja fertilidad, pequeños incrementos pueden producir aumentos significativos en el rendimiento.
>
> Sin embargo, conforme se satisfacen los requerimientos nutricionales del cultivo, el efecto de nuevas aplicaciones disminuye progresivamente.
>
> Este comportamiento refleja uno de los principios fundamentales de la producción agrícola sostenible.

---

> ## 📊 Interpretación desde la Ciencia de Datos

> Este proceso introduce una relación no lineal entre la fertilización y el rendimiento.
>
> Un modelo lineal asume que cada unidad adicional de fertilizante produce siempre el mismo incremento en la producción.
>
> En cambio, el comportamiento simulado refleja que el beneficio marginal disminuye conforme aumenta la fertilización.

# Integración de los procesos agronómicos

Hasta este punto hemos modelado de forma independiente los principales procesos que intervienen en el desarrollo del cultivo.

Sin embargo, el rendimiento agrícola no depende de un único factor.

Es el resultado de la interacción simultánea entre la disponibilidad de agua, las propiedades del suelo, las condiciones climáticas y las prácticas de manejo.

El siguiente paso consiste en integrar todos estos procesos para estimar el potencial productivo del cultivo.

Cada componente aportará una fracción de información al comportamiento final del sistema.

Por esta razón, el rendimiento ya no será calculado mediante una relación lineal simple, sino mediante la combinación de procesos con significado agronómico.

```python
# ==========================================================
# Integración del sistema agrícola
# ==========================================================
#
# En esta etapa integramos todos los procesos modelados
# anteriormente para estimar el potencial productivo del
# cultivo.
#
# Cada componente representa un proceso agronómico con
# significado físico, biológico o de manejo.
#
# El objetivo no consiste en reproducir un cultivo
# específico, sino construir un sistema agrícola coherente
# donde las relaciones entre los factores sean
# interpretables.
# ==========================================================

# ----------------------------------------------------------
# Potencial hídrico
# ----------------------------------------------------------
#
# La disponibilidad de agua es modulada por la eficiencia
# con la que el suelo es capaz de almacenarla y ponerla
# a disposición del cultivo.
# ----------------------------------------------------------

potencial_hidrico = (
    agua_disponible *
    eficiencia_hidrica
)

# ----------------------------------------------------------
# Potencial productivo
# ----------------------------------------------------------
#
# El potencial hídrico es posteriormente modulado por:
#
# • Temperatura
# • pH del suelo
# • Fertilización
# • Radiación
#
# Cada uno representa un proceso independiente que afecta
# el desarrollo fisiológico del cultivo.
# ----------------------------------------------------------

potencial_productivo = (

    potencial_hidrico

    * efecto_temperatura

    * efecto_ph

    * (1 + 0.15 * efecto_fertilizacion)

    + 0.30 * radiacion

)

print("="*65)
print("POTENCIAL PRODUCTIVO CALCULADO")
print("="*65)

print(f"Valor medio : {potencial_productivo.mean():.2f}")
print(f"Valor mínimo: {potencial_productivo.min():.2f}")
print(f"Valor máximo: {potencial_productivo.max():.2f}")
```

# Construcción de modelos predictivos

Después de comprender el comportamiento del conjunto de datos mediante el Análisis Exploratorio de Datos (EDA), el siguiente paso consiste en entrenar diferentes modelos de regresión.

El objetivo no es determinar cuál algoritmo es "mejor" de manera absoluta, sino analizar cómo responde cada uno frente a un fenómeno cuya complejidad ha ido aumentando progresivamente a lo largo de los experimentos.

Se compararán tres enfoques:

- Regresión Lineal
- Random Forest
- XGBoost

La comparación se realizará utilizando exactamente el mismo conjunto de entrenamiento y prueba para garantizar una evaluación objetiva.

```python
# ==========================================================
# Variables predictoras y variable objetivo
# ==========================================================

X = df_exp4[[
    "Precipitacion_mm",
    "Temperatura_C",
    "Radiacion",
    "Materia_Organica_pct",
    "pH",
    "Fertilizacion_kg_ha",
    "Riego_mm"
]]

y = df_exp4["Rendimiento_t_ha"]

print("Variables predictoras:")

display(X.head())

print("\nVariable objetivo:")

display(y.head())
```

```python
# ==========================================================
# División entrenamiento - prueba
# ==========================================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42

)

print("="*60)

print("Entrenamiento:", len(X_train))

print("Prueba:", len(X_test))
```

```python
# ==========================================================
# Regresión Lineal
# ==========================================================

from sklearn.linear_model import LinearRegression

modelo_lr = LinearRegression()

modelo_lr.fit(X_train, y_train)

pred_lr = modelo_lr.predict(X_test)
```

```python
# ==========================================================
# Random Forest
# ==========================================================

from sklearn.ensemble import RandomForestRegressor

modelo_rf = RandomForestRegressor(

    n_estimators=300,

    random_state=42

)

modelo_rf.fit(

    X_train,

    y_train

)

pred_rf = modelo_rf.predict(X_test)
```

```python
# ==========================================================
# XGBoost
# ==========================================================

from xgboost import XGBRegressor

modelo_xgb = XGBRegressor(

    n_estimators=300,

    learning_rate=0.05,

    max_depth=5,

    random_state=42

)

modelo_xgb.fit(

    X_train,

    y_train

)

pred_xgb = modelo_xgb.predict(X_test)
```

# Comparación del desempeño predictivo

Una vez entrenados los modelos, el siguiente paso consiste en evaluar su capacidad para representar el comportamiento del sistema agrícola.

Para garantizar una comparación objetiva, todos los modelos fueron entrenados utilizando exactamente el mismo conjunto de entrenamiento y evaluados sobre el mismo conjunto de prueba.

Se utilizarán tres métricas ampliamente empleadas en problemas de regresión:

- Coeficiente de Determinación (R²)
- Error Cuadrático Medio (RMSE)
- Error Absoluto Medio (MAE)

Estas métricas permitirán analizar cómo evoluciona el desempeño de la Estadística Tradicional y del Machine Learning frente a un fenómeno progresivamente más complejo.

```python
import sklearn

print(sklearn.__version__)
```

```python
# ==========================================================
# Métricas de evaluación
# ==========================================================

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    root_mean_squared_error
)

import pandas as pd

# ------------------------------
# Tabla comparativa
# ------------------------------

resultados = pd.DataFrame({

    "Modelo": [
        "Regresión Lineal",
        "Random Forest",
        "XGBoost"
    ],

    "R²": [
        r2_score(y_test, pred_lr),
        r2_score(y_test, pred_rf),
        r2_score(y_test, pred_xgb)
    ],

    "RMSE": [
        root_mean_squared_error(y_test, pred_lr),
        root_mean_squared_error(y_test, pred_rf),
        root_mean_squared_error(y_test, pred_xgb)
    ],

    "MAE": [
        mean_absolute_error(y_test, pred_lr),
        mean_absolute_error(y_test, pred_rf),
        mean_absolute_error(y_test, pred_xgb)
    ]

})

# Redondear resultados
resultados = resultados.round(4)

# Ordenar por mejor R²
resultados = resultados.sort_values(
    by="R²",
    ascending=False
).reset_index(drop=True)

# Agregar ranking
resultados.insert(0, "Ranking", range(1, len(resultados) + 1))

display(resultados)
```

```python
# ==========================================================
# Ranking de desempeño de los modelos
# ==========================================================

resultados = resultados.sort_values(
    by="R²",
    ascending=False
).reset_index(drop=True)

resultados.index = resultados.index + 1
resultados.index.name = "Ranking"

display(resultados)
```

> ## 📊 Interpretación de resultados
>
> La comparación evidencia cómo el desempeño de los modelos cambia a medida que aumenta la complejidad del fenómeno.
>
> Si el problema mantiene relaciones predominantemente lineales, la Regresión Lineal continúa ofreciendo resultados competitivos.
>
> Sin embargo, cuando aparecen relaciones no lineales e interacciones entre variables, algoritmos como Random Forest y XGBoost comienzan a capturar con mayor precisión la estructura del sistema.

---

> ## 💡 Reflexión metodológica
>
> La elección de un modelo predictivo no debería basarse únicamente en el valor de una métrica.
>
> Debe considerar también la naturaleza del fenómeno, la interpretabilidad requerida y el equilibrio entre simplicidad y capacidad predictiva.

```python
# ==========================================================
# Comparación: Valores Reales vs Predichos
# ==========================================================

import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(18,6))

modelos = [

    ("Regresión Lineal", pred_lr),

    ("Random Forest", pred_rf),

    ("XGBoost", pred_xgb)

]

for ax, (titulo, pred) in zip(axes, modelos):

    ax.scatter(
        y_test,
        pred,
        alpha=0.65
    )

    minimo = min(y_test.min(), pred.min())
    maximo = max(y_test.max(), pred.max())

    ax.plot(
        [minimo, maximo],
        [minimo, maximo],
        linestyle="--",
        linewidth=2
    )

    ax.set_title(titulo)

    ax.set_xlabel("Valores Reales")

    ax.set_ylabel("Valores Predichos")

    ax.grid(alpha=0.30)

plt.tight_layout()

plt.show()
```
